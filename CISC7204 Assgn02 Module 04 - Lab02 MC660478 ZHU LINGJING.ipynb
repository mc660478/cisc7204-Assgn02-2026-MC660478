```python
import pandas as pd
import plotly.express as px
from dash import Dash, html, dcc
```

```python
airline_data = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/airline_data.csv',
                           encoding="ISO-8859-1",
                           dtype={'Div1Airport': str, 'Div1TailNum': str, 
                                  'Div2Airport': str, 'Div2TailNum': str})
```

```python
data = airline_data.sample(n=500, random_state=42)

fig = px.pie(data, values='Flights', names='DistanceGroup', title='Distance group proportion by flight')
```

```python
app = Dash(__name__)
```

```python
app.layout = html.Div(children=[
    # Task 3: Add the application title with CSS styles
    html.H1('Airline Dashboard', 
            style={'textAlign': 'center', 'color': '#503D36', 'font-size': 40}),
    
    # Task 4: Add the application description with CSS styles
    html.P('Proportion of distance group (250 mile distance interval group) by flight', 
           style={'textAlign': 'center', 'color': '#F57241'}),
    
    # Task 5: Embed the Plotly pie chart
    dcc.Graph(figure=fig)
])
```

```python
app.run(debug=True, use_reloader=False, jupyter_mode="tab")
```
