```python
# Import required libraries
import pandas as pd
import plotly.graph_objects as go
import dash
from dash import dcc
from dash import html
from dash.dependencies import Input, Output

# Read the airline data into pandas dataframe
airline_data = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/airline_data.csv',
                           encoding="ISO-8859-1",
                           dtype={'Div1Airport': str, 'Div1TailNum': str, 
                                  'Div2Airport': str, 'Div2TailNum': str})
```

```python
app = dash.Dash(__name__)
```

```python
title = html.H1('Airline Performance Dashboard', 
                style={'textAlign': 'center', 'color': '#503D36', 'font-size': 40})
```

```python
input_box = html.Div(["Input Year: ", dcc.Input(id='input-year', value='2010', type='number', style={'height':'50px', 'font-size': 35})], 
                     style={'font-size': 40})
```

```python
app.layout = html.Div(children=[
    title,
    html.Br(),
    html.Br(),
    input_box,
    html.Br(),
    html.Br(),
    html.Div(dcc.Graph(id='line-plot'))
])
```

```python
@app.callback(
    Output(component_id='line-plot', component_property='figure'),
    Input(component_id='input-year', component_property='value')
)
def get_graph(entered_year):
    df = airline_data[airline_data['Year'] == int(entered_year)]
    line_data = df.groupby('Month')['ArrDelay'].mean().reset_index()
    fig = go.Figure(data=go.Scatter(x=line_data['Month'], y=line_data['ArrDelay'], mode='lines', line_color='green'))
    fig.update_layout(title='Month vs Average Flight Delay Time', xaxis_title='Month', yaxis_title='ArrDelay')
    return fig
```

```python
app.run(debug=True, use_reloader=False, jupyter_mode="tab", port=8051)
```

```python
import plotly.express as px
import dash
from dash import dcc
from dash import html
from dash.dependencies import Input, Output

app2 = dash.Dash(__name__)

app2.layout = html.Div([
    html.H1('Total number of flights to the destination state split by reporting airline', style={'textAlign': 'center', 'color': '#503D36', 'font-size': 40}),
    html.Div(["Input Year: ", dcc.Input(id='input-year-bar', value='2010', type='number', style={'height':'50px', 'font-size': 35})], style={'font-size': 40}),
    html.Br(),
    html.Br(),
    html.Div(dcc.Graph(id='bar-plot'))
])

@app2.callback(
    Output(component_id='bar-plot', component_property='figure'),
    Input(component_id='input-year-bar', component_property='value')
)
def get_bar_graph(entered_year):
    df = airline_data[airline_data['Year'] == int(entered_year)]
    bar_data = df.dropna(subset=['DestState']).groupby(['DestState'])['Flights'].sum().reset_index()
    fig = px.bar(bar_data, x='DestState', y='Flights', title='Flights to Destination State')
    fig.update_layout(xaxis_title='DestState', yaxis_title='Flights')
    return fig
```

```python
app2.run(debug=True, use_reloader=False, jupyter_mode="tab", port=8053)
```
